```python
from pyspark.sql import Row
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# 1. Define the architecture schema
schema = StructType([
    StructField("customer_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("country", StringType(), True),
    StructField("signup_age", IntegerType(), True)
])

# 2. Add sample raw records
raw_data = [
    Row(101, "Alice Smith", "USA", 34),
    Row(102, "Bob Jones", "UK", 45),
    Row(103, "Charlie Singh", "India", 29),
    Row(104, "Diana Müller", "Germany", None) 
]

# 3. Create the data frame in active cluster memory
df_bronze = spark.createDataFrame(raw_data, schema)

# 4. Confirm it exists by printing a preview
display(df_bronze)
```

```python
# Real-World Silver Layer Transformation: Filter and Clean Data
# We filter out any rows where 'signup_age' is Null/None

df_silver = df_bronze.filter(df_bronze.signup_age.isNotNull())

# Let's also add a calculated column to standardize the country text to uppercase
from pyspark.sql.functions import upper
df_silver_clean = df_silver.withColumn("country_code", upper(df_silver.country))

# Display the clean dataset on screen
display(df_silver_clean)
```
